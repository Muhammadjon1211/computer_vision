```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line Plot

```python
x = np.array([1,2,3,4])
y = np.array([10,20,25,30])

plt.plot(x,y)
plt.title("Basic Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Line Styles and Colors

```python
x = np.array([1,2,3,4])
y = np.array([3,8,1,10])

plt.plot(y, linestyle="dotted", color="g", marker="o")
plt.title("Styled Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Multiple Lines Line Graph

```python
x = np.array([1,2,3,4])
y1 = np.array([3,8,1,10])
y2 = np.array([6,2,7,11])

plt.plot(x, y1, label = "line1", color="r", marker="s")
plt.plot(x, y2, label = "line2", color="b", marker="o", linestyle="--")
plt.title("Multiple Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend()
plt.grid()
plt.show()
```

# Bar Chart with Numpy and Matplotlib

```python
x = np.array(["Salary", "Expenses", "Donation", "Charity"])
y = np.array([3, 8, 1, 10])

plt.bar(x,y, color="g")
plt.title("Bar Chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Scatter Plot with Numpy and Matplotlib

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([3,4,5,6,6,9,8,10])
plt.hist(data, bins=5, color="g", edgecolor="black")
plt.title("Histogram")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Subplots (Multiple Graphs Together)

```python
x = np.array([1,2,3,4])
y1 = np.array([3,8,1,10])
y2 = np.array([6,2,7,11])

plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
plt.plot(x, y1, color="r", marker="s")
plt.title("Subplot 1")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.subplot(1,2,2)
plt.bar(x, y2, color="b")
plt.title("Subplot 2")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.tight_layout()
plt.show()
```

```python
x = np.array([1,2,3,4])
y = np.array([4,8,6,10])

plt.plot(x,y, color = "green")
plt.title("Save Figure")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.savefig("my_plot.png")
plt.show()
```
